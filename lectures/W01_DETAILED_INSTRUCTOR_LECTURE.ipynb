# 0:00–0:20 — Why astrostatistics?

## Instructor goal

Students should understand that statistics is not an optional layer placed on top of astronomy. It is the formal language connecting what is measured to what we want to know.

### Start with the inverse problem

Astronomy usually gives us observations such as

$$
D=\{F_i,\ m_i,\ z_i,\ t_i,\sigma_i,\ldots\}.
$$

But the scientific targets are frequently latent:

$$
\theta=
\{H_0,\Omega_m,w,\tau,L,M,\ldots\}.
$$

The telescope does not directly return a column labelled

$$
\Omega_m=0.31.
$$

Instead, a physical model predicts how latent parameters generate observables.

### Forward versus inverse direction

Write on the board:

$$
\boxed{
\theta
\longrightarrow
\text{physical model}
\longrightarrow
\text{observables}
\longrightarrow
D_{\rm obs}
}
$$

and then underneath:

$$
\boxed{
D_{\rm obs}
\longrightarrow
\text{statistical inference}
\longrightarrow
p(\theta\mid D)
}
$$

The first arrow is the **generative / forward problem**.  
The second is the **inverse / inferential problem**.

### Example: supernova cosmology

A simplified forward chain is

$$
(H_0,\Omega_m)
\rightarrow
H(z)
\rightarrow
D_L(z)
\rightarrow
\mu(z)
\rightarrow
m_B
\rightarrow
m_{B,\rm obs}.
$$

### Example: reverberation mapping

A simplified AGN chain is

$$
C(t)
\rightarrow
L(t)
=
\int \Psi(\tau)C(t-\tau)d\tau
\rightarrow
\{F_j,t_j,\sigma_j\}.
$$

### Ask the room

**Question 1:** Which quantities in these chains are measured?  
**Question 2:** Which are inferred?  
**Question 3:** Where does uncertainty enter?

Do not yet correct every answer. The point is to expose the distinction that Lab 01 will make explicit.

## A useful course principle

Throughout the semester, ask five questions before running an algorithm:

1. **What is the estimand?**
2. **What is actually observed?**
3. **What model connects the two?**
4. **What uncertainty or selection acts on the observations?**
5. **What result would make us distrust our analysis?**

This is more important than memorizing the syntax of `scipy`, `emcee`, or `sklearn`.

# 0:20–0:50 — Probability calculus

The Crete notebook begins with the basic probability axioms.

## Event probability

For an event \(A\),

$$
0\le P(A)\le 1.
$$

The complement is

$$
P(A^c)=1-P(A).
$$

If \(A,B,C,\ldots\) are mutually exclusive and exhaustive,

$$
P(A)+P(B)+P(C)+\cdots=1.
$$

### Astronomy examples

- \(A\): an observed photon falls inside a source aperture.
- \(A\): an AGN belongs to a chosen redshift bin.
- \(A\): a candidate passes a quality cut.
- \(A\): a lag estimate lies inside a specified interval.

The event must always be defined with respect to a sample space.

### Instructor prompt

Ask:

> “What is the sample space if I say the probability that a supernova has \(z>0.5\)?”

Students should realize that the answer depends on what population or observed sample is being referenced.

This anticipates selection effects in Week 6.

## Probability is conditional on the setup

A statement such as

$$
P(z>0.5)
$$

is incomplete unless we know **for what population**.

For example,

$$
P(z>0.5\mid \text{Pantheon+})
$$

is an empirical statement about the Pantheon+ compilation.

It is not automatically equal to

$$
P(z>0.5\mid \text{all SNe Ia in the Universe}).
$$

This difference between observed sample and parent population will recur throughout the course.

# 0:50–1:15 — Conditional probability, factorization, independence and marginalization

This section follows the order of the Crete `Intro.ipynb`.

## Conditional probability

$$
P(A\mid B)
$$

means the probability of \(A\) conditional on \(B\).

The joint probability can be factorized:

$$
P(A,B)=P(A\mid B)P(B).
$$

Equivalently,

$$
P(A,B)=P(B\mid A)P(A).
$$

Therefore,

$$
P(A\mid B)
=
\frac{P(B\mid A)P(A)}{P(B)}.
$$

This is already the algebraic structure of Bayes' theorem, although formal Bayesian inference comes in Week 4.

### Astronomy interpretation

Let

- \(A\): object is an AGN;
- \(B\): object has a particular color.

Then

$$
P(A\mid B)
$$

is not the same quantity as

$$
P(B\mid A).
$$

This distinction matters greatly in classification.

## Independence

Events \(A\) and \(B\) are independent if

$$
P(A\mid B)=P(A),
$$

which implies

$$
P(A,B)=P(A)P(B).
$$

### Critical astronomy example

Two catalogue rows are not necessarily independent just because they appear on different lines of a table.

If the same SN is measured by two surveys, those rows can share:

- the same physical source;
- calibration uncertainty;
- peculiar-velocity uncertainty;
- systematic terms.

Similarly, multiple epochs from one AGN are temporally correlated.

So:

$$
\boxed{
\text{different rows}\neq\text{independent draws}.
}
$$

This point directly motivates the Lab 01 repeated-`CID` exercise and later covariance treatment.

## Marginalization

Suppose \(B_i\) are mutually exclusive possibilities. Then

$$
P(A)=\sum_iP(A,B_i).
$$

For a continuous nuisance variable \(b\),

$$
p(A)=\int p(A,b)\,db.
$$

This operation becomes central in Bayesian inference.

For example, if cosmological inference depends on a nuisance calibration \(M_B\),

$$
p(\Omega_m\mid D)
=
\int
p(\Omega_m,M_B\mid D)\,dM_B.
$$

### Instructor remark

Tell students explicitly:

> “The marginalization formula you learn today is the same operation we will use in Week 4 to remove a nuisance calibration parameter.”

# 1:15–1:30 — Continuous variables and probability density

The Crete notebook moves from probability of events to continuous variables using an astronomical color example.

For a continuous random variable \(X\),

$$
P(X=x)=0
$$

for any exact value \(x\), in the mathematical sense.

What is meaningful is an interval:

$$
P(a<X<b)
=
\int_a^b f(x)\,dx.
$$

The probability density satisfies

$$
f(x)\ge 0
$$

and

$$
\int_{-\infty}^{\infty}f(x)\,dx=1.
$$

### Important unit point

A density has units inverse to its variable.

If \(x\) is in magnitudes, then \(f(x)\) has units of inverse magnitudes.

Therefore a density value such as

$$
f(x)=2
$$

does not violate probability axioms. The **integral** over an interval must be between 0 and 1.

## Example: redshift

Imagine a normalized redshift density \(p(z)\).

Then

$$
P(0.2<z<0.5)
=
\int_{0.2}^{0.5}p(z)\,dz.
$$

The probability of obtaining exactly

$$
z=0.314159\ldots
$$

is zero for an ideal continuous distribution.

### Connection to histograms

A histogram approximates density only after accounting for bin width and normalization.

This is why changing binning can alter visual impressions even when the underlying sample is unchanged.

# 1:40–2:00 — Gaussian distributions, CDFs and intervals

## Gaussian distribution

The Crete notebook introduces

$$
f(x)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp
\left[
-\frac{(x-\mu)^2}{2\sigma^2}
\right].
$$

Parameters:

- \(\mu\): mean/location;
- \(\sigma\): standard deviation/scale.

In astronomy Gaussian models are common because of measurement-noise approximations and the central limit theorem—but not every astronomical distribution is Gaussian.

### Ask students for non-Gaussian examples

Possible answers:

- flux-limited samples;
- positive-only quantities;
- luminosity functions;
- time-lag posteriors;
- eccentricities;
- outlier-contaminated residuals.

## Cumulative distribution function

For continuous \(X\),

$$
F(x)
=
P(X\le x)
=
\int_{-\infty}^{x}f(x')dx'.
$$

The CDF is dimensionless and ranges from 0 to 1.

Quantiles are obtained by inverting the CDF.

For example, the median \(m\) satisfies

$$
F(m)=0.5.
$$

### Central interval versus confidence interval

If a theoretical distribution has central 68% interval

$$
[q_{0.16},q_{0.84}],
$$

that is a probability interval of the distribution.

It is not automatically a frequentist confidence interval for an unknown parameter.

This wording distinction matters later.

## Mean, median and mode

For a symmetric Gaussian,

$$
\text{mean}=\text{median}=\text{mode}.
$$

For asymmetric distributions these can differ substantially.

This is relevant to:

- skewed posteriors;
- lag distributions;
- censored samples;
- selection-truncated observables.

### Instructor prompt

Ask:

> “If a lag posterior is bimodal, is reporting only the mean scientifically useful?”

The desired answer is: often not.

# 2:00–2:20 — Discrete variables, PMFs and Poisson counts

The Crete notebook next distinguishes continuous densities from discrete probability mass functions.

For a discrete variable \(K\),

$$
P(K=k)
$$

is a genuine probability mass.

Normalization is

$$
\sum_{k=0}^{\infty}P(K=k)=1.
$$

## Poisson distribution

For counts occurring independently at a constant expected rate \(\lambda\),

$$
P(K=k\mid\lambda)
=
\frac{\lambda^k e^{-\lambda}}{k!}.
$$

The mean and variance are both

$$
E[K]=\lambda,
\qquad
{\rm Var}(K)=\lambda.
$$

### Astronomy use

Photon counts can be Poisson-like when detector/background complications are neglected.

This is why Gaussian errors can fail in the low-count regime.

## Continuous versus discrete: why the distinction matters

For a flux measurement we may write a density

$$
p(F_{\rm obs}\mid F_{\rm true}).
$$

For a photon count we may write a probability mass

$$
P(K=k\mid\lambda).
$$

Using the wrong sampling model changes the likelihood and therefore changes inference.

This will become central when students construct likelihoods in Week 3.

# 2:40–2:55 — Generative models for cosmology and AGN variability

## Supernova chain

A schematic model is

$$
\theta_{\rm cosmo}
\rightarrow
D_L(z;\theta_{\rm cosmo})
\rightarrow
\mu(z)
\rightarrow
m_B^{\rm true}
\rightarrow
m_B^{\rm obs}.
$$

Potential nuisance variables include:

- absolute calibration;
- host-galaxy effects;
- peculiar velocities;
- survey calibration;
- selection.

## AGN chain

A continuum process \(C(t)\) drives a line response:

$$
L(t)
=
\int\Psi(\tau)C(t-\tau)d\tau.
$$

The survey then samples these continuous processes at irregular times:

$$
\{t_i,C_i,\sigma_{C,i}\},
\qquad
\{t_j,L_j,\sigma_{L,j}\}.
$$

The observations are not the underlying continuous processes.
