## 0:00–0:35 — Why MCMC?

Grid evaluation scales exponentially with parameter dimension.

MCMC constructs a correlated sequence

$$
\theta^{(1)},\theta^{(2)},\ldots
$$

whose stationary distribution is the target posterior.

The purpose is integration/exploration of a probability distribution—not “finding the best fit.”

## 0:35–1:10 — Markov chains and convergence

Introduce:

- stationarity;
- burn-in/warm-up;
- autocorrelation;
- integrated autocorrelation time;
- effective sample size;
- acceptance fraction.

A chain of \(10^5\) samples is not equivalent to \(10^5\) independent samples.

Roughly,

$$
N_{\rm eff}\sim\frac{N}{\tau_{\rm int}}.
$$

## 1:10–1:30 — Diagnostics before corner plots

Trace plots reveal:

- drift;
- poor mixing;
- stuck walkers;
- multiple modes.

A plausible corner plot can still be generated from an inadequate chain.

Therefore:

$$
\boxed{\text{diagnose first, summarize second}.}
$$

## 1:40–2:15 — Covariance in cosmology

For correlated data,

$$
\chi^2
=
\Delta^TC^{-1}\Delta.
$$

A systematic effect can correlate many SNe simultaneously. It therefore cannot generally be represented by simply increasing each diagonal error independently.

Compare:

$$
C_{\rm stat}
$$

and

$$
C_{\rm stat+sys}.
$$

## 2:15–2:40 — Numerical linear algebra

Avoid explicitly calculating \(C^{-1}\) when possible.

If

$$
C=LL^T
$$

is a Cholesky decomposition, solve linear systems using \(L\).

This is more stable and generally faster than forming the inverse.

## 2:40–3:00 — Validation strategy

Because the teaching posterior has only two parameters, the grid posterior is known.

This lets students test whether MCMC reproduces:

- posterior median;
- interval width;
- correlation.

That is an unusually valuable validation exercise.

### Bridge to Lab 05

Students compare statistical and full systematic covariance, run MCMC, inspect traces/autocorrelation, deliberately run a bad chain, and verify the sampler against the grid posterior.

## End-of-lecture synthesis

Before the lab, students should be able to answer:

1. **What is the estimand?**
2. **What are the actual observations?**
3. **What stochastic assumptions connect the two?**
4. **Where can calibration, covariance or selection enter?**
5. **What diagnostic would make the result untrustworthy?**

The lab then turns these questions into executable real-data analysis.
