## 0:00–0:35 — Bayes' theorem

$$
p(\theta\mid D)
=
\frac{p(D\mid\theta)p(\theta)}{p(D)}.
$$

Identify:

- likelihood \(p(D\mid\theta)\);
- prior \(p(\theta)\);
- posterior \(p(\theta\mid D)\);
- evidence \(p(D)\).

In parameter estimation, the evidence is a normalization constant unless we compare models.

## 0:35–1:10 — Priors are part of the model

Discuss:

- bounded uniform priors;
- Gaussian informative priors;
- scale parameters;
- truncation;
- prior support;
- physically impossible regions.

A posterior is not “the data speaking alone.” It is conditional on:

$$
\text{data}+\text{likelihood}+\text{prior}+\text{model}.
$$

## 1:10–1:30 — Marginalization

For nuisance parameter \(\phi\),

$$
p(\theta\mid D)
=
\int p(\theta,\phi\mid D)\,d\phi.
$$

In a discrete grid,

$$
p(\theta_i\mid D)\propto \sum_j p(\theta_i,\phi_j\mid D).
$$

Connect this directly to the \((\Omega_m,M_B)\) posterior.

## 1:30–1:40 — Break

## 1:40–2:10 — Credible regions and posterior geometry

Explain:

- marginal interval;
- joint credible region;
- posterior correlation;
- highest-density region versus equal-tail interval.

A 68% posterior contour should enclose 68% of posterior probability, not merely correspond to an arbitrary contour level.

## 2:10–2:35 — Prior sensitivity

A scientifically mature Bayesian analysis asks:

> Would a defensible alternative prior materially change the conclusion?

Students compare a broad prior on \(\Omega_m\) with a deliberately informative teaching prior.

If posterior width collapses under an informative prior, the tighter result may reflect prior information rather than data information.

## 2:35–3:00 — Posterior predictive thinking

Posterior-supported parameter values imply a family of predicted Hubble relations.

A posterior predictive check asks whether model predictions resemble relevant features of the observed data.

It is a **model criticism tool**, not another parameter estimator.

### Bridge to Lab 04

Students normalize a real posterior grid, marginalize, calculate credible intervals, test prior sensitivity and inspect posterior-supported Hubble relations.

## End-of-lecture synthesis

Before the lab, students should be able to answer:

1. **What is the estimand?**
2. **What are the actual observations?**
3. **What stochastic assumptions connect the two?**
4. **Where can calibration, covariance or selection enter?**
5. **What diagnostic would make the result untrustworthy?**

The lab then turns these questions into executable real-data analysis.
