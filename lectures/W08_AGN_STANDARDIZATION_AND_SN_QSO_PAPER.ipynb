## 0:00–0:35 — Standard candle versus standardisable candle

A standard candle has known luminosity.

A standardisable candle uses an empirical relationship to infer luminosity or distance after calibration.

For an AGN relation such as

$$
\log L_X=\gamma\log L_{\rm UV}+\beta,
$$

the cosmological utility depends on:

- intrinsic scatter;
- redshift evolution;
- selection;
- calibration;
- independence of the relation from the cosmology being tested.

## 0:35–1:05 — Flux, luminosity and cosmology

$$
L=4\pi D_L^2F.
$$

Therefore a luminosity-based relation can become a distance relation, but only if its calibration is appropriately handled.

Ask continuously:

> Which quantity is directly observed?

> Which quantity was computed using an assumed distance?

## 1:05–1:30 — Joint SN–QSO likelihood

For approximately independent datasets,

$$
\ln\mathcal L_{\rm joint}
=
\ln\mathcal L_{\rm SN}
+
\ln\mathcal L_{\rm QSO}.
$$

Discuss nuisance calibration offsets and the paper's parameterization.

Do not interpret a joint likelihood as proof that the probes are physically independent in every systematic respect.

## 1:30–1:40 — Break

## 1:40–2:15 — Statistical selection versus physical standardisation

A photometrically “central” AGN in a multivariate feature distribution is not automatically a physically standardized distance indicator.

Distinguish:

$$
\text{statistical representativeness}
$$

from

$$
\text{physical standardisability}.
$$

## 2:15–2:40 — Circularity and double use of information

If objects are selected partly because they align with a posterior-calibrated Hubble relation, that same alignment is not independent evidence for the cosmology.

This is a general statistical issue:

$$
\text{selection on outcome-related information}
\rightarrow
\text{optimistic apparent agreement}.
$$

## 2:40–3:00 — Critical paper-reading checklist

Students identify:

- direct observations;
- derived quantities;
- calibration parameters;
- cosmological parameters;
- selection criteria;
- validation claims;
- limitations stated by the authors.

## End-of-lecture synthesis

Before the lab, students should be able to answer:

1. **What is the estimand?**
2. **What are the actual observations?**
3. **What stochastic assumptions connect the two?**
4. **Where can calibration, covariance or selection enter?**
5. **What diagnostic would make the result untrustworthy?**

The lab then turns these questions into executable real-data analysis.
