## 0:00–0:40 — Selection is part of the likelihood

If an object enters a catalogue only when it passes a detection/selection event \(S\),

$$
p(D\mid\theta,S)
$$

is not generally the same as

$$
p(D\mid\theta).
$$

For a selection-corrected likelihood, normalization may depend on

$$
P(S\mid\theta).
$$

Ignoring that dependence can bias population or cosmological inference.

## 0:40–1:15 — Truncation versus censoring

Distinguish:

- truncation: objects outside a criterion are absent;
- censoring: an object is known to exist but only a bound is observed;
- missingness;
- quality cuts;
- survey-specific targeting.

Connect to flux limits and Malmquist-type effects.

## 1:15–1:30 — Robustness is not selection correction

A useful real-data sensitivity exercise is to repeat inference after:

- changing \(z_{\max}\);
- dropping high-uncertainty objects;
- omitting one survey;
- splitting low/high redshift.

But:

$$
\boxed{\text{sensitivity analysis}\neq\text{formal selection-function correction}.}
$$

## 1:40–2:15 — Influence and leverage

A point can be:

- statistically unusual;
- high leverage;
- highly influential;
- perfectly valid.

Do not remove an observation merely because it changes the result.

Use influence analysis to understand fragility.

## 2:15–2:40 — Survey heterogeneity and domain shift

Different surveys may occupy different regions in:

- redshift;
- signal-to-noise;
- calibration;
- cadence;
- selection.

This creates a precursor to the machine-learning concept of domain shift.

## 2:40–3:00 — Designing a robustness statement

A good conclusion distinguishes:

- stable under tested perturbations;
- unstable under tested perturbations;
- formally corrected for selection.

Only the first two are established by Lab 06.

### Bridge to Lab 06

Students repeatedly fit the same real-data model under defensible sample changes and quantify shifts rather than declaring one subset “correct.”

## End-of-lecture synthesis

Before the lab, students should be able to answer:

1. **What is the estimand?**
2. **What are the actual observations?**
3. **What stochastic assumptions connect the two?**
4. **Where can calibration, covariance or selection enter?**
5. **What diagnostic would make the result untrustworthy?**

The lab then turns these questions into executable real-data analysis.
