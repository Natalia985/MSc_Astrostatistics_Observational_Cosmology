## 0:00–0:35 — Joint likelihoods

If datasets are conditionally independent given parameters,

$$
p(D_1,D_2\mid\theta)
=
p(D_1\mid\theta)p(D_2\mid\theta).
$$

Therefore

$$
\ln\mathcal L_{\rm joint}
=
\ln\mathcal L_1+\ln\mathcal L_2.
$$

But independence is an assumption. Shared calibration or modelling systematics can invalidate naive multiplication.

## 0:35–1:10 — Complementarity

Two probes can have broad individual posteriors but constrain different directions in parameter space.

Geometrically, intersecting degeneracy directions can yield a much tighter joint region.

This is more informative than saying simply:

> “More data gives smaller error bars.”

## 1:10–1:30 — BAO conceptually

BAO provides a standard-ruler type constraint involving combinations such as

$$
D_M(z)/r_d
$$

and

$$
D_H(z)/r_d.
$$

These probe expansion differently from luminosity distances.

Students use published summary products rather than reducing spectra.

## 1:30–1:40 — Break

## 1:40–2:10 — Probe-specific nuisance parameters

Each probe has its own calibration/nuisance structure.

Examples:

- SN: absolute magnitude / distance-ladder calibration;
- QSO/AGN: empirical relation calibration and intrinsic scatter;
- BAO: sound-horizon dependence or chosen compressed observable.

A joint analysis must keep these roles explicit.

## 2:10–2:35 — Tension, consistency and model dependence

Differences between probe posteriors can reflect:

- statistical fluctuation;
- systematics;
- selection;
- incompatible calibrations;
- inadequate cosmological model.

Do not label every discrepancy a “tension” without quantifying it.

## 2:35–3:00 — Course synthesis

Return to the semester's chain:

$$
\text{real observations}
\rightarrow
\text{data model}
\rightarrow
\text{likelihood}
\rightarrow
\text{posterior}
\rightarrow
\text{diagnostics}
\rightarrow
\text{robust scientific claim}.
$$

The most important skill is not running a sampler. It is being able to defend every arrow in that chain.

### Bridge to Lab 11 and final project

Students combine real probes, inspect each probe separately, perform one robustness perturbation and state what the joint result gains—and what it still cannot establish.

## End-of-lecture synthesis

Before the lab, students should be able to answer:

1. **What is the estimand?**
2. **What are the actual observations?**
3. **What stochastic assumptions connect the two?**
4. **Where can calibration, covariance or selection enter?**
5. **What diagnostic would make the result untrustworthy?**

The lab then turns these questions into executable real-data analysis.
