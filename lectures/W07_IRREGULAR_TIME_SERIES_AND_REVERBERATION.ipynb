## 0:00–0:35 — The reverberation problem

Continuum variations drive a delayed line response:

$$
L(t)=\int \Psi(\tau)C(t-\tau)\,d\tau.
$$

In a simplified narrow-response picture,

$$
L(t)\approx A\,C(t-\tau).
$$

The estimand is the lag \(\tau\), but the observations are noisy, irregular samples of \(C\) and \(L\).

## 0:35–1:05 — Irregular sampling

Discuss:

- unequal cadence;
- seasonal gaps;
- finite baseline;
- heteroscedastic errors;
- missing epochs.

A time lag cannot be assessed independently of the window function and baseline.

## 1:05–1:30 — Correlation-based lag estimation

Introduce:

- discrete correlation ideas;
- interpolation cross-correlation (ICCF);
- peak versus centroid lag;
- lag-search window.

The search window is part of the analysis design. It should not be tuned after seeing the desired result.

## 1:30–1:40 — Break

## 1:40–2:15 — Red noise and stochastic variability

AGN variability is temporally correlated.

A simple DRW/OU covariance can be written schematically as

$$
k(\Delta t)
=
\sigma^2\exp(-|\Delta t|/\tau_{\rm DRW}).
$$

Correlated stochastic variability changes the null behavior of apparent cross-correlation features.

## 2:15–2:40 — Gaussian processes

A GP specifies

$$
f(t)\sim\mathcal{GP}[m(t),k(t,t')].
$$

Given irregular observations, the GP provides a joint predictive distribution at unobserved times.

Stress:

- interpolation is uncertain;
- the kernel encodes assumptions;
- a smooth-looking GP curve is not observed data.

## 2:40–3:00 — Lag uncertainty and validation

Discuss:

- FR/RSS-style resampling;
- jackknife/sensitivity;
- alternative search windows;
- multimodal lag distributions;
- comparison to a published benchmark only after one's own inference.

Rest-frame conversion:

$$
\tau_{\rm rest}=\frac{\tau_{\rm obs}}{1+z}.
$$

### Bridge to Lab 07

Students work with extracted SDSS-RM light curves—not spectra—and must decide whether a lag is scientifically usable rather than merely return a number.

## End-of-lecture synthesis

Before the lab, students should be able to answer:

1. **What is the estimand?**
2. **What are the actual observations?**
3. **What stochastic assumptions connect the two?**
4. **Where can calibration, covariance or selection enter?**
5. **What diagnostic would make the result untrustworthy?**

The lab then turns these questions into executable real-data analysis.
