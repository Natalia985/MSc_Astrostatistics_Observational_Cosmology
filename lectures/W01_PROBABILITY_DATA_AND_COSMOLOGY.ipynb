## 0:00–0:20 — Why astrostatistics?

Astronomy is fundamentally an inverse problem. We observe photons, fluxes, magnitudes, redshifts and time series, but usually want latent physical quantities.

A useful distinction is

$$
\text{forward model}: \theta \rightarrow D,
$$

versus

$$
\text{inference}: D \rightarrow p(\theta\mid D).
$$

For observational cosmology,

$$
\theta_{\rm cosmo}=(H_0,\Omega_m,w,\ldots)
$$

generates a cosmic expansion history, which generates distances, which generate observables.

The statistical problem is to reverse that chain while respecting measurement uncertainty, calibration and selection.

## 0:20–1:05 — Probability calculus

Review:

$$
P(A),\qquad P(A^c)=1-P(A),
$$

$$
P(A,B)=P(A\mid B)P(B),
$$

and marginalization

$$
P(A)=\sum_i P(A,B_i).
$$

For continuous variables, probability is represented through a density:

$$
P(a<X<b)=\int_a^b p(x)\,dx.
$$

Emphasize that a PDF value is **not itself a probability**.

### Cosmology connection

A likelihood such as

$$
p(D\mid \Omega_m)
$$

is a probability model for observations conditional on a cosmological parameter. It is not the same object as

$$
p(\Omega_m\mid D).
$$

That distinction will become central in Week 4.

## 1:05–1:25 — Random variables and astronomical measurements

Discuss:

- discrete versus continuous variables;
- PMF, PDF and CDF;
- Gaussian and asymmetric distributions;
- mean, median and mode;
- quantiles.

Use real astronomical examples:

- photon counts: discrete;
- redshift: continuous;
- survey identifier: categorical;
- light-curve flux: continuous;
- quality flag: discrete/categorical.

### Important language

A central 68% interval of a known probability distribution is **not automatically a frequentist confidence interval for an unknown parameter**.

## 2:10–2:40 — Generative thinking

A minimal supernova generative chain is

$$
(H_0,\Omega_m)
\rightarrow D_L(z)
\rightarrow \mu(z)
\rightarrow m_B
\rightarrow m_{B,\rm obs}.
$$

A minimal reverberation chain is

$$
C(t)
\rightarrow L(t)=\int \Psi(\tau)C(t-\tau)\,d\tau
\rightarrow \{F_j,\sigma_j,t_j\}.
$$

The same statistical discipline applies to both:

- define latent quantities;
- define observations;
- define noise;
- define dependencies;
- define selection.

## End-of-lecture synthesis

Before the lab, students should be able to answer:

1. **What is the estimand?**
2. **What are the actual observations?**
3. **What stochastic assumptions connect the two?**
4. **Where can calibration, covariance or selection enter?**
5. **What diagnostic would make the result untrustworthy?**

The lab then turns these questions into executable real-data analysis.
