## 0:00–0:35 — Population, sample and estimator

Distinguish:

- population parameter;
- finite observed sample;
- estimator;
- sampling distribution.

If

$$
\hat\theta=T(D),
$$

then \(\hat\theta\) varies across hypothetical repetitions of the data-acquisition process.

This is why a point estimate without uncertainty is incomplete.

## 0:35–1:15 — Bootstrap

For observed data

$$
x_1,\ldots,x_N,
$$

a nonparametric bootstrap repeatedly samples \(N\) observations with replacement.

For the median,

$$
\tilde x^{*(b)}
=
{\rm median}(x_{i_1}^{*(b)},\ldots,x_{i_N}^{*(b)}).
$$

The empirical distribution of \(\tilde x^{*(b)}\) approximates the estimator's sampling variability.

### Astronomy warning

The resampling unit must match the independent observational unit.

If one SN appears in multiple catalogue rows, blindly bootstrapping rows can over-count one physical object.

## 1:15–1:35 — Effect size before significance

For two groups, define an interpretable quantity such as

$$
\Delta_{\rm med}
=
{\rm median}(r_{\rm high-z})
-
{\rm median}(r_{\rm low-z}).
$$

Then quantify uncertainty on \(\Delta_{\rm med}\).

The scientific question is not merely whether a null can be rejected; it is also:

> How large is the effect, and is that magnitude scientifically consequential?

## 1:45–2:20 — Permutation tests

Suppose the null is that group labels are exchangeable.

Compute

$$
T_{\rm obs}
=
\left|
{\rm median}(r_2)-{\rm median}(r_1)
\right|.
$$

Shuffle labels repeatedly, preserving group sizes, to generate \(T^{*(b)}\).

Estimate

$$
p
=
\frac{1+\sum_b I(T^{*(b)}\ge T_{\rm obs})}{B+1}.
$$

Explain what this probability is—and what it is not.

It is not

$$
P(H_0\mid D).
$$

## 2:20–2:40 — Robust statistics and ECDFs

Review:

- mean versus median;
- standard deviation versus MAD;
- empirical CDF;
- tail sensitivity;
- effect of outliers.

An ECDF avoids arbitrary histogram-bin choices:

$$
\widehat F(x_{(i)})=\frac{i}{N}.
$$

## 2:40–3:00 — Researcher degrees of freedom

Discuss:

- choosing a redshift split after seeing the result;
- repeated testing;
- deleting influential points because they weaken a result;
- HARKing / post-hoc hypotheses.

The lab therefore **pre-specifies** a split and later tests modest sensitivity to that choice.

## End-of-lecture synthesis

Before the lab, students should be able to answer:

1. **What is the estimand?**
2. **What are the actual observations?**
3. **What stochastic assumptions connect the two?**
4. **Where can calibration, covariance or selection enter?**
5. **What diagnostic would make the result untrustworthy?**

The lab then turns these questions into executable real-data analysis.
