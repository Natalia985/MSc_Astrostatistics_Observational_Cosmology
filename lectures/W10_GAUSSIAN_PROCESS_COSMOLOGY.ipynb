## 0:00–0:35 — Parametric versus non-parametric inference

Parametric cosmology assumes a form such as flat \(\Lambda\)CDM:

$$
D_L(z;\Omega_m,H_0).
$$

A GP reconstruction instead places a prior over functions:

$$
f(z)\sim\mathcal{GP}[m(z),k(z,z')].
$$

This can test flexibility of the data-model relation without immediately assigning every deviation to new physics.

## 0:35–1:10 — GP conditioning

Given training inputs \(X\), observations \(\mathbf y\), covariance \(K\) and noise covariance \(\Sigma\), the predictive mean is

$$
\mu_* =
K_{*X}(K_{XX}+\Sigma)^{-1}\mathbf y.
$$

The predictive covariance is

$$
K_{**}
-
K_{*X}(K_{XX}+\Sigma)^{-1}K_{X*}.
$$

The uncertainty is part of the result.

## 1:10–1:30 — Kernels encode assumptions

Examples:

- squared exponential: very smooth;
- Matérn: rougher functions;
- sums/products: structured behavior.

A kernel is not a neutral interpolation choice.

## 1:40–2:15 — Hyperparameters and evidence

Length scale controls how quickly the function varies.

Too large:

$$
\text{over-smoothing}.
$$

Too small:

$$
\text{over-flexibility / fitting noise}.
$$

Discuss optimization versus marginalization of GP hyperparameters.

## 2:15–2:40 — Covariance and correlated observations

If the observational covariance is non-diagonal, the GP noise model should reflect it where feasible.

Binning can make a teaching exercise computationally manageable, but binning is itself a data transformation and must be documented.

## 2:40–3:00 — Cosmological interpretation

A GP deviation from a reference \(\Lambda\)CDM curve is not automatically evidence for exotic cosmology.

Possible origins include:

- calibration;
- covariance misspecification;
- kernel choice;
- redshift-dependent systematics;
- genuine model inadequacy.

### Bridge to Lab 10

Students compare GP reconstructions under at least two kernels and must diagnose over- versus under-smoothing.

## End-of-lecture synthesis

Before the lab, students should be able to answer:

1. **What is the estimand?**
2. **What are the actual observations?**
3. **What stochastic assumptions connect the two?**
4. **Where can calibration, covariance or selection enter?**
5. **What diagnostic would make the result untrustworthy?**

The lab then turns these questions into executable real-data analysis.
