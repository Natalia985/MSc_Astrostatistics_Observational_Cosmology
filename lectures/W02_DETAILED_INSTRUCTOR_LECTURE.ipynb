# 0:00–0:25 — Statistics themselves are random variables

The Crete notebook begins with a useful construction: if

$$
X\sim \mathcal N(\mu,\sigma^2)
$$

and

$$
Y\sim \mathcal N(\mu,\sigma^2)
$$

are independent, then their sum

$$
Z=X+Y
$$

is also Gaussian:

$$
Z\sim\mathcal N(2\mu,2\sigma^2).
$$

Therefore

$$
E[Z]=2\mu,
\qquad
{\rm Var}(Z)=2\sigma^2.
$$

### Why begin here?

Because a statistic computed from random data is itself random.

Examples:

- sample mean;
- sample median;
- maximum;
- correlation;
- fitted cosmological parameter.

If we repeated the observational process, the statistic would change.

This motivates the **sampling distribution**.

## Astronomical translation

Suppose we compute the median Hubble residual from a finite SN sample:

$$
\tilde r=T(D).
$$

Even if the underlying Universe and survey mechanism were unchanged, a different observed set of SNe would produce another value of \(\tilde r\).

Therefore we need not only

$$
\hat\theta
$$

but also uncertainty on \(\hat\theta\).

### Instructor prompt

Ask:

> “If two analyses give medians 0.01 mag and 0.03 mag, can we say they are different?”

Students should answer: not without the sampling uncertainty.

# 0:25–0:55 — Null hypothesis, statistic and p-value

The Crete notebook emphasizes a model-checking question:

> Does a theoretical model produce data/statistics compatible with what we observed?

The structure is:

1. define a null hypothesis \(H_0\);
2. choose a statistic \(T\);
3. derive or estimate the distribution of \(T\) under \(H_0\);
4. compare observed \(T_{\rm obs}\) to that distribution.

## The p-value

For an appropriate tail definition,

$$
p
=
P(T\ \text{at least as extreme as}\ T_{\rm obs}\mid H_0).
$$

### Crucial interpretation

A p-value is **not**

$$
P(H_0\mid D).
$$

It does not give the probability that the model is true.

It also does not “accept” the model when large.

The Crete notebook makes this point explicitly: failure to reject one statistic does not validate the complete model.

## Significance level

Before evaluating data, choose a decision threshold \(\alpha\), for example

$$
\alpha=0.05.
$$

If

$$
p<\alpha,
$$

we reject \(H_0\) according to the chosen rule.

### But scientific importance is separate

A minuscule effect can have a small p-value when \(N\) is enormous.

A substantial effect can have a large p-value when \(N\) is small.

Therefore always accompany significance with an **effect size**.

## Hubble-residual example

Suppose

$$
r_i=\mu_i-\mu_{\rm ref}(z_i).
$$

Define

$$
\Delta_{\rm med}
=
{\rm median}(r_{\rm high-z})
-
{\rm median}(r_{\rm low-z}).
$$

This is an effect size.

Then a test can ask whether a difference at least this large would be unusual under a null model.

The order matters:

$$
\boxed{
\text{define effect}
\rightarrow
\text{quantify uncertainty}
\rightarrow
\text{test}
}
$$

# 0:55–1:15 — Standard error, Z tests and t tests

The Crete notebook uses the sample mean as a test statistic.

For independent observations with population standard deviation \(\sigma\),

$$
{\rm SE}(\bar X)
=
\frac{\sigma}{\sqrt N}.
$$

Then

$$
Z
=
\frac{\bar X-\mu_0}{\sigma/\sqrt N}.
$$

If \(\sigma\) is unknown and estimated by sample standard deviation \(s\), use

$$
t
=
\frac{\bar X-\mu_0}{s/\sqrt N},
$$

with

$$
\nu=N-1
$$

degrees of freedom under the classical assumptions.

### Why the t distribution?

Estimating \(\sigma\) introduces extra uncertainty, particularly for small \(N\).

The t distribution has heavier tails than a standard Gaussian.

## Assumptions matter

The simple standard-error formula assumes independent observations.

If observations are correlated,

$$
{\rm Var}(\bar X)
$$

contains covariance terms.

For correlated measurements,

$$
{\rm Var}(\bar X)
=
\frac{1}{N^2}
\sum_{i,j}{\rm Cov}(X_i,X_j).
$$

So the familiar

$$
\sigma/\sqrt N
$$

can underestimate uncertainty.

This is a direct preview of why Pantheon+ supplies a covariance matrix.

# 1:15–1:30 — Central Limit Theorem

Under appropriate conditions, the distribution of the normalized sum/mean tends toward a Gaussian as sample size increases:

$$
\frac{\bar X-\mu}{\sigma/\sqrt N}
\Rightarrow
\mathcal N(0,1).
$$

### What the CLT does not say

It does not say:

- the original data become Gaussian;
- all estimators are Gaussian;
- small samples are safe;
- strong dependence can be ignored;
- heavy tails never matter.

### Astronomy example

A non-Gaussian distribution of individual SN residuals can still yield an approximately Gaussian sampling distribution for the mean if conditions are sufficiently favorable.

# 1:40–2:00 — Distribution tests

The Crete notebook discusses normality tests and the Kolmogorov–Smirnov test.

## K–S statistic

For empirical CDF \(F_N(x)\) and reference CDF \(F_0(x)\),

$$
D
=
\sup_x
|F_N(x)-F_0(x)|.
$$

For two samples,

$$
D
=
\sup_x
|F_{N_1}(x)-G_{N_2}(x)|.
$$

### Why ECDF thinking is useful

Histograms depend on:

- bin edges;
- bin width;
- normalization.

ECDFs avoid these choices.

### Important caution

Different tests are sensitive to different properties.

A test of the mean and a K–S test need not produce the same p-value because they are asking different questions.

## Normality testing caution

With very large \(N\), tiny deviations from Gaussianity can become “statistically significant.”

The relevant scientific question may instead be:

> Are deviations from Gaussianity large enough to materially affect the inferential method?

This is why diagnostics should be tied to consequences, not only test thresholds.

# 2:00–2:25 — Non-parametric testing and permutation logic

The Crete notebook introduces non-parametric testing by constructing the sampling distribution of a statistic under a null model.

For two observed groups \(A\) and \(B\), a permutation test uses exchangeability.

Observed statistic:

$$
T_{\rm obs}
=
\left|
{\rm median}(B)-{\rm median}(A)
\right|.
$$

Under the null that labels carry no information, pool observations and repeatedly permute group labels.

For permutation \(b\),

$$
T^{(b)}
=
\left|
{\rm median}(B^{(b)})
-
{\rm median}(A^{(b)})
\right|.
$$

Then estimate

$$
p
=
\frac{
1+\sum_b I(T^{(b)}\ge T_{\rm obs})
}{
B+1
}.
$$

## What is being assumed?

A permutation test is not “assumption free.”

It assumes the observations are exchangeable under the null with respect to the permutation being performed.

If observations have clustering, repeated objects or time dependence, arbitrary row permutation may be invalid.

Again:

$$
\boxed{
\text{the unit of analysis matters}.
}
$$

# 2:25–2:45 — Bootstrap uncertainty

Bootstrap and permutation address different questions.

## Bootstrap

Approximate the sampling distribution of an estimator.

Observed sample:

$$
x_1,\ldots,x_N.
$$

Bootstrap sample:

$$
x_1^*,\ldots,x_N^*
$$

drawn with replacement from the observed sample.

For each resample compute

$$
\hat\theta^{*(b)}.
$$

The collection

$$
\{\hat\theta^{*(1)},\ldots,\hat\theta^{*(B)}\}
$$

approximates the estimator's sampling variation.

## Bootstrap versus permutation

### Bootstrap asks

> If I repeatedly observed similar samples, how variable would my estimator be?

### Permutation asks

> Under a null of exchangeability, how often would I see a difference this extreme?

The same dataset can support both analyses, but they answer different questions.

### Lab 02 application

Students will estimate

$$
\Delta_{\rm med}
$$

and then:

- bootstrap it for uncertainty;
- permutation-test it against a null.

# 2:45–2:55 — Multiple testing, researcher degrees of freedom and robustness

Suppose we try redshift splits at

$$
z=0.1,0.15,0.2,\ldots,1.0
$$

and report only the split with the smallest p-value.

The nominal p-value no longer represents the full search procedure.

This is a form of **researcher degrees of freedom** / multiple testing.

### Better practice

Pre-specify a primary comparison.

Then use nearby alternatives as **sensitivity analyses**, not as opportunities to hunt significance.

Similarly, do not remove outliers only because they weaken the desired conclusion.

Instead ask:

> How much does the result depend on influential observations?
