## 0:00–0:35 — From a probability model to a likelihood

Given data \(D\) and parameter \(\theta\),

$$
\mathcal L(\theta;D)=p(D\mid\theta).
$$

The data are fixed; the likelihood is regarded as a function of the parameter.

For independent Gaussian observations,

$$
-2\ln\mathcal L
=
\sum_i\frac{(y_i-\mu_i(\theta))^2}{\sigma_i^2}
+\text{constant}.
$$

But the Pantheon+ data are correlated, so the appropriate quadratic form is

$$
\chi^2
=
\Delta^T C^{-1}\Delta.
$$

## 0:35–1:10 — Flat-\(\Lambda\)CDM forward model

For

$$
E(z)=\sqrt{\Omega_m(1+z)^3+1-\Omega_m},
$$

$$
H(z)=H_0E(z).
$$

Then

$$
D_L(z)=
(1+z)\frac{c}{H_0}
\int_0^z\frac{dz'}{E(z')}.
$$

Distance modulus:

$$
\mu(z)=5\log_{10}\left(\frac{D_L}{\mathrm{Mpc}}\right)+25.
$$

For standardized SN brightness,

$$
m_B^{\rm corr}
=
\mu(z)+M_B+\epsilon.
$$

## 1:10–1:30 — Identifiability and nuisance parameters

Because

$$
D_L\propto H_0^{-1},
$$

changing \(H_0\) shifts the distance-modulus zero point.

That shift is degenerate with \(M_B\).

Therefore SN apparent magnitudes alone do not separately identify both \(H_0\) and \(M_B\) without external calibration.

This is a model-identifiability issue, not an optimizer failure.

## 1:40–2:10 — Optimization

Discuss:

- grid search;
- gradient-free optimization;
- local versus global optimum;
- parameter boundaries;
- profiling nuisance parameters.

For a nuisance parameter entering linearly,

$$
\chi^2=a-2M_Bb+M_B^2d,
$$

so

$$
\widehat M_B=\frac{b}{d}.
$$

This allows students to profile \(M_B\) analytically while scanning \(\Omega_m\).

## 2:10–2:35 — Likelihood geometry

Define

$$
\Delta\chi^2(\theta)
=
\chi^2(\theta)-\chi^2_{\min}.
$$

Use the profile curve to understand:

- curvature;
- parameter sensitivity;
- local uncertainty;
- boundaries.

Stress that standard \(\Delta\chi^2\) interval interpretations require regularity assumptions.

## 2:35–3:00 — Diagnostics

A good optimization result still requires model checking.

Inspect residuals versus:

- redshift;
- survey;
- uncertainty;
- other relevant covariates.

### Bridge to Lab 03

Students use the published Pantheon+ statistical covariance, not just diagonal plotting errors, and explicitly test the \(H_0-M_B\) zero-point degeneracy.

## End-of-lecture synthesis

Before the lab, students should be able to answer:

1. **What is the estimand?**
2. **What are the actual observations?**
3. **What stochastic assumptions connect the two?**
4. **Where can calibration, covariance or selection enter?**
5. **What diagnostic would make the result untrustworthy?**

The lab then turns these questions into executable real-data analysis.
