## 0:00–0:35 — Multivariate distance

A diagonal standardized distance is

$$
D_{\rm diag}^2
=
\sum_j
\left(
\frac{x_j-\tilde x_j}{s_j}
\right)^2.
$$

It treats standardized coordinates as if their covariance were diagonal.

If features are correlated, use the Mahalanobis form:

$$
D_M^2
=
(\mathbf x-\boldsymbol\mu)^TC^{-1}(\mathbf x-\boldsymbol\mu).
$$

## 0:35–1:05 — Why colors are correlated

If features include both magnitudes and colors such as

$$
u,g,r,i,z,
$$

and

$$
u-g,\quad g-r,\quad r-i,\quad i-z,
$$

the feature vector contains algebraically dependent information.

This can make covariance ill-conditioned and makes “ten independent dimensions” an incorrect interpretation.

## 1:05–1:30 — PCA

After standardization, PCA diagonalizes the sample covariance:

$$
C=V\Lambda V^T.
$$

Principal components are directions of maximal sample variance—not automatically physically meaningful latent variables.

Use loadings and explained variance to interpret what drives the feature space.

## 1:40–2:10 — Train/test leakage

If multiple measurements of one AGN exist, random row splitting allows the same physical object to appear in training and test sets.

Similarly, if a selection criterion incorporates a cosmology-calibrated residual, evaluating that same residual on the selected sample is not an independent test.

Leakage is fundamentally about **information pathways**, not just software mistakes.

## 2:10–2:35 — Calibration and trustworthy ML

For classifiers or probabilistic scores, assess:

- discrimination;
- calibration;
- class imbalance;
- grouped cross-validation;
- domain shift.

A high accuracy alone is insufficient.

## 2:35–3:00 — Robust centers and covariance regularization

Discuss:

- median/MAD;
- robust covariance;
- shrinkage;
- condition number;
- pseudo-inverse versus regularized inverse.

## End-of-lecture synthesis

Before the lab, students should be able to answer:

1. **What is the estimand?**
2. **What are the actual observations?**
3. **What stochastic assumptions connect the two?**
4. **Where can calibration, covariance or selection enter?**
5. **What diagnostic would make the result untrustworthy?**

The lab then turns these questions into executable real-data analysis.
